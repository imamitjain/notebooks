```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

sns.set_theme(style="whitegrid")
%matplotlib inline
```

## 1. Line Plots and Scatter Plots

```python
x = np.linspace(0, 10, 100)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Line plot
axes[0].plot(x, np.sin(x), label='sin(x)', linewidth=2)
axes[0].plot(x, np.cos(x), label='cos(x)', linewidth=2, linestyle='--')
axes[0].set_xlabel('x')
axes[0].set_ylabel('y')
axes[0].set_title('Line Plot')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Scatter plot
rng = np.random.default_rng(42)
x_scatter = rng.standard_normal(100)
y_scatter = 2 * x_scatter + rng.standard_normal(100) * 0.5
colors = rng.uniform(0, 1, 100)
axes[1].scatter(x_scatter, y_scatter, c=colors, cmap='viridis', alpha=0.7, edgecolors='k', linewidth=0.5)
axes[1].set_xlabel('x')
axes[1].set_ylabel('y')
axes[1].set_title('Scatter Plot')

plt.tight_layout()
plt.show()
```

## 2. Histograms and Distributions

```python
rng = np.random.default_rng(42)
normal_data = rng.standard_normal(1000)
uniform_data = rng.uniform(-3, 3, 1000)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].hist(normal_data, bins=40, edgecolor='black', alpha=0.7, color='steelblue')
axes[0].set_title('Normal Distribution')
axes[0].set_xlabel('Value')
axes[0].set_ylabel('Frequency')

# KDE plot with seaborn
sns.kdeplot(data=normal_data, ax=axes[1], fill=True, label='Normal')
sns.kdeplot(data=uniform_data, ax=axes[1], fill=True, label='Uniform')
axes[1].set_title('KDE Comparison')
axes[1].legend()

plt.tight_layout()
plt.show()
```

## 3. Bar Charts and Count Plots

```python
categories = ['Python', 'JavaScript', 'Go', 'Rust', 'Java']
popularity = [85, 78, 45, 35, 70]

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].bar(categories, popularity, color=sns.color_palette('muted', len(categories)), edgecolor='black')
axes[0].set_title('Language Popularity')
axes[0].set_ylabel('Score')

# Seaborn countplot
tips = sns.load_dataset('tips')
sns.countplot(data=tips, x='day', hue='sex', ax=axes[1])
axes[1].set_title('Tips by Day and Gender')

plt.tight_layout()
plt.show()
```

## 4. Heatmaps and Correlation Matrices

```python
rng = np.random.default_rng(42)
data = pd.DataFrame(rng.standard_normal((100, 5)), columns=['A', 'B', 'C', 'D', 'E'])
data['B'] = data['A'] * 0.8 + rng.standard_normal(100) * 0.3
data['D'] = -data['C'] * 0.6 + rng.standard_normal(100) * 0.4

corr = data.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0,
            square=True, linewidths=1)
plt.title('Correlation Matrix')
plt.tight_layout()
plt.show()
```

## 5. Subplots and Figure Layout

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
rng = np.random.default_rng(42)

# Top-left: line
x = np.linspace(0, 5, 50)
axes[0, 0].plot(x, x**2, 'b-', label='x²')
axes[0, 0].plot(x, x**3 / 10, 'r--', label='x³/10')
axes[0, 0].legend()
axes[0, 0].set_title('Polynomials')

# Top-right: scatter
axes[0, 1].scatter(rng.standard_normal(50), rng.standard_normal(50), alpha=0.6)
axes[0, 1].set_title('Random Scatter')

# Bottom-left: bar
axes[1, 0].barh(['A', 'B', 'C', 'D'], [30, 45, 20, 55], color='teal')
axes[1, 0].set_title('Horizontal Bars')

# Bottom-right: histogram
axes[1, 1].hist(rng.standard_normal(500), bins=30, edgecolor='black', alpha=0.7)
axes[1, 1].set_title('Distribution')

fig.suptitle('Subplot Grid Demo', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 6. Seaborn Statistical Plots (boxplot, violinplot, pairplot)

```python
tips = sns.load_dataset('tips')

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

sns.boxplot(data=tips, x='day', y='total_bill', hue='sex', ax=axes[0])
axes[0].set_title('Box Plot')

sns.violinplot(data=tips, x='day', y='total_bill', ax=axes[1], inner='quartile')
axes[1].set_title('Violin Plot')

sns.stripplot(data=tips, x='day', y='total_bill', hue='sex', dodge=True, alpha=0.6, ax=axes[2])
axes[2].set_title('Strip Plot')

plt.tight_layout()
plt.show()
```
