## 1. Array Creation

```python
# From lists
a = np.array([1, 2, 3, 4, 5])
print(f"1D array: {a}, shape: {a.shape}, dtype: {a.dtype}")

# 2D array
matrix = np.array([[1, 2, 3], [4, 5, 6]])
print(f"2D array shape: {matrix.shape}")

# Common constructors
zeros = np.zeros((3, 4))
ones = np.ones((2, 3))
identity = np.eye(3)
rng = np.random.default_rng(42)
random_arr = rng.standard_normal((3, 3))

print(f"Zeros:\n{zeros}")
print(f"Identity:\n{identity}")
print(f"Random:\n{random_arr}")
```

## 2. Indexing and Slicing

```python
arr = np.arange(20).reshape(4, 5)
print(f"Array:\n{arr}\n")

# Basic slicing
print(f"Row 1: {arr[1]}")
print(f"Column 2: {arr[:, 2]}")
print(f"Submatrix (2x2 top-left):\n{arr[:2, :2]}")

# Boolean indexing
mask = arr > 10
print(f"\nElements > 10: {arr[mask]}")

# Fancy indexing
rows = [0, 2, 3]
print(f"Rows 0, 2, 3:\n{arr[rows]}")
```

## 3. Broadcasting

Broadcasting lets NumPy operate on arrays of different shapes without copying data.

```python
# Scalar broadcast
a = np.array([[1, 2, 3], [4, 5, 6]])
print(f"a * 10:\n{a * 10}\n")

# Row vector broadcast
row = np.array([100, 200, 300])
print(f"a + row:\n{a + row}\n")

# Column vector broadcast
col = np.array([[10], [20]])
print(f"a + col:\n{a + col}")
```

## 4. Linear Algebra (ML Essentials)

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

# Matrix multiplication
print(f"A @ B:\n{A @ B}\n")

# Transpose
print(f"A.T:\n{A.T}\n")

# Determinant and inverse
print(f"det(A): {np.linalg.det(A):.1f}")
print(f"inv(A):\n{np.linalg.inv(A)}\n")

# Eigenvalues (used in PCA, spectral methods)
eigenvalues, eigenvectors = np.linalg.eig(A)
print(f"Eigenvalues: {eigenvalues}")
```

## 5. Vectorized Operations (Why NumPy is Fast)

```python
import time

size = 1_000_000
py_list = list(range(size))
np_arr = np.arange(size)

# Python loop
start = time.perf_counter()
result_py = [x ** 2 for x in py_list]
time_py = time.perf_counter() - start

# NumPy vectorized
start = time.perf_counter()
result_np = np_arr ** 2
time_np = time.perf_counter() - start

print(f"Python loop: {time_py:.4f}s")
print(f"NumPy:       {time_np:.4f}s")
print(f"Speedup:     {time_py / time_np:.0f}x")
```
