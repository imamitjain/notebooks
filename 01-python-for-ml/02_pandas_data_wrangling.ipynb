```python
import pandas as pd
import numpy as np

print(f"Pandas version: {pd.__version__}")
```

## 1. DataFrames and Series

```python
# Creating Series and DataFrames
s = pd.Series([10, 20, 30, 40], index=['a', 'b', 'c', 'd'])
print(f"Series:\n{s}\n")

df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Carol', 'Dave', 'Eve'],
    'age': [25, 30, 35, 28, 32],
    'city': ['NYC', 'LA', 'NYC', 'Chicago', 'LA'],
    'salary': [70000, 80000, 90000, 75000, 85000]
})
print(df)
print(f"\nShape: {df.shape}")
print(f"Columns: {list(df.columns)}")
print(f"Dtypes:\n{df.dtypes}")
```

## 2. Loading Data (CSV, JSON, Parquet)

```python
# Create sample data and save/load in different formats
df.to_csv('sample.csv', index=False)
df_csv = pd.read_csv('sample.csv')
print("From CSV:")
print(df_csv.head())

df.to_json('sample.json', orient='records')
df_json = pd.read_json('sample.json')
print("\nFrom JSON:")
print(df_json.head())
```

## 3. Exploration and Summary Statistics

```python
print(df.describe())
print(f"\nInfo:")
df.info()
print(f"\nValue counts for city:\n{df['city'].value_counts()}")
print(f"\nMean salary: ${df['salary'].mean():,.0f}")
print(f"Median age: {df['age'].median()}")
```

## 4. Filtering, Sorting, and Selection

```python
# Boolean filtering
high_salary = df[df['salary'] > 75000]
print("Salary > 75k:")
print(high_salary)

# Multiple conditions
nyc_senior = df[(df['city'] == 'NYC') & (df['age'] > 30)]
print(f"\nNYC & age > 30:\n{nyc_senior}")

# Sorting
sorted_df = df.sort_values('salary', ascending=False)
print(f"\nSorted by salary:\n{sorted_df}")

# .loc and .iloc
print(f"\n.loc[0:2, 'name':'age']:\n{df.loc[0:2, 'name':'age']}")
print(f"\n.iloc[0:2, 0:2]:\n{df.iloc[0:2, 0:2]}")
```

## 5. Handling Missing Data

```python
# Create data with missing values
df_missing = pd.DataFrame({
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 2, 3, np.nan, 5],
    'C': [1, 2, 3, 4, 5]
})
print("Original:")
print(df_missing)
print(f"\nNull counts:\n{df_missing.isnull().sum()}")

# Drop rows with any NaN
print(f"\nDropna:\n{df_missing.dropna()}")

# Fill with mean
filled = df_missing.fillna(df_missing.mean())
print(f"\nFilled with mean:\n{filled}")

# Forward fill
print(f"\nForward fill:\n{df_missing.ffill()}")
```

## 6. GroupBy and Aggregation

```python
# Expand the dataset
rng = np.random.default_rng(42)
sales = pd.DataFrame({
    'product': rng.choice(['Widget', 'Gadget', 'Doohickey'], 100),
    'region': rng.choice(['North', 'South', 'East', 'West'], 100),
    'quantity': rng.integers(1, 50, 100),
    'price': rng.uniform(10, 100, 100).round(2)
})
sales['revenue'] = sales['quantity'] * sales['price']

# Group by product
print("Revenue by product:")
print(sales.groupby('product')['revenue'].agg(['sum', 'mean', 'count']))

# Group by multiple columns
print("\nMean quantity by product and region:")
print(sales.groupby(['product', 'region'])['quantity'].mean().unstack())
```

## 7. Merging and Joining

```python
orders = pd.DataFrame({
    'order_id': [1, 2, 3, 4, 5],
    'customer_id': [101, 102, 101, 103, 102],
    'amount': [250, 150, 300, 200, 175]
})

customers = pd.DataFrame({
    'customer_id': [101, 102, 103, 104],
    'name': ['Alice', 'Bob', 'Carol', 'Dave'],
    'tier': ['Gold', 'Silver', 'Gold', 'Bronze']
})

# Inner join
merged = pd.merge(orders, customers, on='customer_id', how='inner')
print("Inner merge:")
print(merged)

# Left join (keeps all orders)
left = pd.merge(orders, customers, on='customer_id', how='left')
print(f"\nLeft merge:\n{left}")

# Total spend per customer
spend = merged.groupby('name')['amount'].sum().sort_values(ascending=False)
print(f"\nTotal spend:\n{spend}")
```
