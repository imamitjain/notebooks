## 1. Variables, Types, and Zero Values

```go
import "fmt"

// Short declaration
x := 42
name := "gopher"
pi := 3.14
active := true

fmt.Printf("x=%d, name=%s, pi=%.2f, active=%v\n", x, name, pi, active)

// Zero values - Go initializes everything
var i int
var s string
var b bool
var f float64
fmt.Printf("Zero values: int=%d, string=%q, bool=%v, float=%f\n", i, s, b, f)
```

## 2. Slices — Go's Dynamic Arrays

```go
import "fmt"

// Create and manipulate slices
nums := []int{1, 2, 3, 4, 5}
fmt.Printf("nums: %v, len=%d, cap=%d\n", nums, len(nums), cap(nums))

// Append
nums = append(nums, 6, 7)
fmt.Printf("after append: %v\n", nums)

// Slicing (like Python but no negative indices)
fmt.Printf("nums[2:5]: %v\n", nums[2:5])
fmt.Printf("nums[:3]:  %v\n", nums[:3])

// Make with pre-allocated capacity
big := make([]int, 0, 100)
fmt.Printf("pre-allocated: len=%d, cap=%d\n", len(big), cap(big))
```

## 3. Maps

```go
import "fmt"

// Map creation
scores := map[string]int{
    "alice": 95,
    "bob":   87,
    "carol": 92,
}

// Access and check existence
score, exists := scores["alice"]
fmt.Printf("alice: %d (exists: %v)\n", score, exists)

_, exists = scores["dave"]
fmt.Printf("dave exists: %v\n", exists)

// Iterate
for name, score := range scores {
    fmt.Printf("  %s: %d\n", name, score)
}
```

## 4. Structs and Methods

```go
import "fmt"
import "math"

type Point struct {
    X, Y float64
}

// Value receiver
func (p Point) Distance(other Point) float64 {
    dx := p.X - other.X
    dy := p.Y - other.Y
    return math.Sqrt(dx*dx + dy*dy)
}

// Pointer receiver (can modify the struct)
func (p *Point) Scale(factor float64) {
    p.X *= factor
    p.Y *= factor
}

a := Point{3, 4}
b := Point{0, 0}
fmt.Printf("Distance from %v to %v: %.2f\n", a, b, a.Distance(b))

a.Scale(2)
fmt.Printf("After Scale(2): %v\n", a)
```

## 5. Interfaces

```go
import "fmt"
import "math"

type Shape interface {
    Area() float64
    String() string
}

type Circle struct {
    Radius float64
}

func (c Circle) Area() float64 {
    return math.Pi * c.Radius * c.Radius
}

func (c Circle) String() string {
    return fmt.Sprintf("Circle(r=%.1f)", c.Radius)
}

type Rectangle struct {
    Width, Height float64
}

func (r Rectangle) Area() float64 {
    return r.Width * r.Height
}

func (r Rectangle) String() string {
    return fmt.Sprintf("Rect(%.1fx%.1f)", r.Width, r.Height)
}

// Function accepting the interface
func printArea(s Shape) {
    fmt.Printf("%s -> area = %.2f\n", s, s.Area())
}

printArea(Circle{Radius: 5})
printArea(Rectangle{Width: 3, Height: 4})
```

## 6. Error Handling

```go
import (
    "errors"
    "fmt"
    "strconv"
)

// Functions return errors as values
func divide(a, b float64) (float64, error) {
    if b == 0 {
        return 0, errors.New("division by zero")
    }
    return a / b, nil
}

result, err := divide(10, 3)
if err != nil {
    fmt.Println("Error:", err)
} else {
    fmt.Printf("10 / 3 = %.4f\n", result)
}

_, err = divide(10, 0)
if err != nil {
    fmt.Println("Error:", err)
}

// Parsing with error handling
val, err := strconv.Atoi("not_a_number")
if err != nil {
    fmt.Printf("Parse error: %v (val=%d)\n", err, val)
}
```
